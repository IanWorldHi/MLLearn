```python
#Recurral Network example RNN
x = torch.randn(1, 10) #the input
prev_h = torch.randn(1, 20) #hidden state of rnn that gives it its memory
w_x = torch.randn(20, 10, requires_grad=True) #learning input for hidden state
#requires_grad enables gradient tracking for .backwards
w_h = torch.randn(20, 20, requires_grad=True) #learning weight for input
#MM = matrix multiplication
#.t() is the transpose of a matrix (swaps the dimensions) (why the transpose?)
i2h = torch.mm(w_x, x.t())
h2h = torch.mm(w_h, prev_h.t())
next_h = i2h + h2h #adding output of two matrix multiplications together
next_h = next_h.tanh() #applying activation function, hyperbolic tangent
loss = next_h.sum() #loss function, difference between correct output and predicted output
loss.backward() #backpropagation, smth to do with gradients, derivatives or smth

```
