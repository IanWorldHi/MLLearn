```python
#Tensor manipulation
RT = torch.rand(2, 3)
RT, RT+2, RT*2, RT-2, RT/2, RT**2, torch.log(RT), torch.sin(RT) #(cos, tan, etc)
```

```python
#Matrix multiplication refresher
RT=torch.rand(2, 3)
RT2=torch.rand(3, 2)
RT3=torch.rand(2, 3)
#doing RT*RT2 wouldn't work, is element wise
torch.matmul(RT, RT2), torch.matmul(RT, torch.transpose(RT3, 0, 1)) #for 2D can do .mm, RT3.T for 2D
#RT.permute for full control of order of tranpose dimensions when larger than 3D 
```

```python
#Aggregiation
RT=torch.rand(2, 3)
#Both work, but torch. is more common, also only torch.max(...) can do two tensor comparision
#Can specify torch.mean(x.type(float32))
RT, RT.max(), torch.max(RT), RT.mean(), torch.mean(RT), RT.sum(), torch.sum(RT), RT.max(dim=0), torch.max(RT, dim=0), torch.argmax(RT) #index
```
