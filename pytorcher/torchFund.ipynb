```python
#Often, machine starts with randomized tensors then optimizes
RANDTENSOR = torch.rand(1, 9, 4)
RANDTENSOR2 = torch.rand(1, 9, 4, dtype=torch.float16, device=None, requires_grad=False)
#defaults to none (CPU), gradiant means operations tracked for optimization 
ZEROTENSOR = torch.zeros(1, 9, 4)
ZEROTENSOR2 = torch.zeros_like(RANDTENSOR2)
RANGTENSOR = torch.arange(1, 10, 0.5) #start, end, step amount
RANDTENSOR.dtype, ZEROTENSOR.dtype, ZEROTENSOR2.dtype, RANDTENSOR2.device, RANDTENSOR2.requires_grad
```

```python
#Tensor manipulation
RT = torch.rand(2, 3)
RT, RT+2, RT*2, RT-2, RT/2, RT**2, torch.log(RT), torch.sin(RT) #(cos, tan, etc)
```

```python
#Matrix multiplication refresher
RT=torch.rand(2, 3)
RT2=torch.rand(3, 2)
RT3=torch.rand(2, 3)
#doing RT*RT2 wouldn't work, is element wise
torch.matmul(RT, RT2), torch.matmul(RT, torch.transpose(RT3, 0, 1)) #for 2D can do .mm, RT3.T for 2D
#RT.permute for full control of order of tranpose dimensions when larger than 3D 
```

```python
#Aggregiation
RT=torch.rand(2, 3)
#Both work, but torch. is more common, also only torch.max(...) can do two tensor comparision
#Can specify torch.mean(x.type(float32))
RT, RT.max(), torch.max(RT), RT.mean(), torch.mean(RT), RT.sum(), torch.sum(RT), RT.max(dim=0), torch.max(RT, dim=0), torch.argmax(RT) #index
```

```python
#Reshape, Stacking, Squeezing, Unsqueezing, change tensor type
T2 = torch.tensor([[1, 4, 5]]) 
T1 = torch.tensor([[1, 2, 3], [1, 4, 5]], dtype=torch.float16) 
T1 = T1.type(torch.float32)
TRESHAPE = T1.reshape(3, 2)  #(T1.reshape(3, -1))Must have same amount of elements, -1 for automatic size only for 1 dimension of it - can also use .view but worse only for contiguous & changes original tensor
TSTACK = torch.stack([T1, T1, T1, T1], dim=1) #same shape tensors to stack
T3 = T2.squeeze() #removes dimensions of size 1, can specify dims
T4 = T2.unsqueeze(dim=0) #adds dimension of size 1, can specify dims
T1.dtype, T1, TRESHAPE, TSTACK, T2, T3, T4
```

```python
#Indexing
T1 = torch.tensor([[[1, 2, 3], [1, 4, 5]], [[4, 2, 4], [1, 4, 9]]]) 
#Syntax: "[x:y:z]" (start:stop:step), so [:] means everything, [1:6:2] means 1, 3, 5
T1.shape, T1, T1[0][0], T1[0, 0], T1[:, 1], T1[:, 0, :], T1[:, 0, 0], T1[0][0][1:2:1]
```

```python
#Pytorch Numpy interaction
array = np.array([1, 2, 3]) 
tensor = torch.from_numpy(array) #Note, makes copy, not same memory
array2 = tensor.numpy()
array, tensor, array2
```
