## Creating Tensors

Tensors are like arrays/matrices to encode inputs/outputs.
- Can auto infer type
- Default is float32
- Can run on GPU or CPU

```python
# Create tensor from Python list
data = [[1, 2], [1, 3]]
torchdata = torch.tensor(data)
print("Tensor from list:")
print(torchdata)
```

```python
# Create tensor from NumPy array
nparr = np.array(data)
torcharr = torch.from_numpy(nparr)

print(f"Shape: {torcharr.shape}")
print(f"Data type: {torcharr.dtype}")
print(f"Device: {torcharr.device}")
print(torcharr)
```

## Tensor Operations

Docs: https://pytorch.org/docs/stable/torch.html

```python
# Create tensor of ones with specific dtype and device
ones_tensor = torch.ones((5, 3), dtype=torch.float64, device='cpu')
print("Tensor of ones:")
print(ones_tensor)
```

```python
# Random tensors with reproducibility
torch.manual_seed(42)  # Sets seed for reproducibility
rand1 = torch.rand(2, 4)
print("Random tensor:")
print(rand1)
```
